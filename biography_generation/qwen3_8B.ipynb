In this notebook you will find the code used to run baseline experiment with Qwen3. The outputs are saved in a JSON file in: ```outputs/baseline/```

This code can be run using an MLX quantization and full models with transformers.

````python
# mlx imports -- use to work with MLX, using quantized (4bit) models
import mlx.core as mx
from utils.mlx_utils import load_model_mlx, generate_biography_mlx

# transformers imports -- use to work with torch, using non-quantized models
#import torch
#from huggingface_hub import login
#from utils.torch_utils import load_model, generate_biography

# common imports
from config import QWEN3, QWEN3_PARAMS
from utils.common import save_response
from prompts import PROMPTS
````

````python
# transformer run token import of the models
#TOKEN = ""
#login(TOKEN)
````

````python
# Selection of prompt used
prompts = ["v1_short", "v2_short", "v3_short"]

# Seed for reproducibility
SEED = 42

# mxl:
mx.random.seed(SEED)

# transformers:
#torch.manual_seed = SEED

# Load model
model, tokenizer = load_model_mlx(QWEN3)
````

````python
# Itterate over all prompts selected
for prompt in prompts: 
    
    prompt_set = PROMPTS[prompt]
    system_prompt = prompt_set["system"]
    user_prompt_nodis = prompt_set["no_dis"]
    user_prompt_withdis = prompt_set["with_dis"]

    # Change number of iteration to have more or less biographies.
    for iteration_number in range(3):
    # Generate no-disability prompt biography
        biography_no_dis = generate_biography_mlx(
            model = model,
            tokenizer = tokenizer,
            user_prompt = user_prompt_nodis,
            system_prompt = system_prompt,
            params = QWEN3_PARAMS,
            enable_thinking=False # Change to True or remove to enable thinking of Qwen model. However this will generate more token so max_token needs to be changed in utils.mlx_utils.generate_biography_mlx.py to avoid mid-answer cuts.
        )

        save_response(data=biography_no_dis,
                filepath="outputs/baseline/results_qwen.json",
                prompt_set=prompt,
                variant="nodis",
                system_prompt=system_prompt,
                user_prompt=user_prompt_nodis,
                iteration_number=iteration_number,
                technique="mlx"
                )

    for iteration_number in range(3):
        # Generate with-disability prompt biography
        biography_with_dis = generate_biography_mlx(
            model = model,
            tokenizer = tokenizer,
            user_prompt = user_prompt_withdis,
            system_prompt = system_prompt,
            params = QWEN3_PARAMS,
            enable_thinking=False
        )
        # Save results
        save_response(data=biography_with_dis,
                filepath="outputs/baseline/results_qwen.json",
                prompt_set=prompt,
                variant="withdis",
                system_prompt=system_prompt,
                user_prompt=user_prompt_withdis,
                iteration_number=iteration_number,
                technique="mlx"
                )
````
