```python
import re
import pandas as pd
from pathlib import Path
```

```python
BASE_DIR = Path().resolve()
GENDER_DIR = BASE_DIR / "resources/gender"
EVALUATION_GENDER_DIR = GENDER_DIR / "gender_detection_evaluation"

GENDER_COL = "genre_auto"
DOC_RE = re.compile(
    r"^(?P<model>[^_]+)_(?P<prompt>v\d+)_(?P<length>[^_]+)_"
    r"(?P<disability>[^_]+)_(?P<run>\d+)"
)
```

```python
df = pd.read_csv(EVALUATION_GENDER_DIR / "gender_detection_benchmark.csv")
df[["model", "prompt"]] = df["doc_id"].str.split("_", expand=True)[[0, 1]]
```

```python
df.assign(fem=df.annotation.eq("Fem")).pivot_table(
    index="model", columns="prompt", values="fem", aggfunc="mean"
).mul(100).round(1)
```
